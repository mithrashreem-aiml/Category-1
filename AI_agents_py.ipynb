```python
class SimpleReflexAgent:
    def __init__(self):
        pass

    def act(self, location, status):
        """
        location: Current location ('A' or 'B')
        status: 'Dirty' or 'Clean'
        """
        if status == "Dirty":
            return "Suck"
        elif location == "A":
            return "Move Right"
        elif location == "B":
            return "Move Left"
agent = SimpleReflexAgent()

test_cases = [
    ("A", "Dirty"),
    ("A", "Clean"),
    ("B", "Dirty"),
    ("B", "Clean")
]

for location, status in test_cases:
    action = agent.act(location, status)
    print(f"Location: {location}, Status: {status} -> Action: {action}")
```

```python
# Simple Reflex Agent - Vacuum Cleaner

def reflex_agent(location, status):
    if status == "Dirty":
        return "Suck"
    elif location == "A":
        return "Move Right"
    else:
        return "Move Left"

# Main Program
location = input("Enter current location (A/B): ")
status = input("Enter room status (Clean/Dirty): ")

action = reflex_agent(location, status)

print("Action:", action)
```

```python
# Model-Based Agent - Door Controller

class ModelBasedAgent:
    def __init__(self):
        self.door_state = "Closed"   # Internal model

    def act(self, command):
        if command == "Open":
            if self.door_state == "Closed":
                self.door_state = "Open"
                return "Door Opened"
            else:
                return "Door is already Open"

        elif command == "Close":
            if self.door_state == "Open":
                self.door_state = "Closed"
                return "Door Closed"
            else:
                return "Door is already Closed"

        else:
            return "Invalid Command"

agent = ModelBasedAgent()

while True:
    command = input("Enter command (Open/Close/Exit): ")

    if command == "Exit":
        break

    print(agent.act(command))
```

```python
# Model-Based Agent - Light Control

class LightAgent:
    def __init__(self):
        self.light_state = "OFF"   # Internal memory

    def act(self, command):
        if command == "ON":
            if self.light_state == "OFF":
                self.light_state = "ON"
                return "Light Turned ON"
            else:
                return "Light is already ON"

        elif command == "OFF":
            if self.light_state == "ON":
                self.light_state = "OFF"
                return "Light Turned OFF"
            else:
                return "Light is already OFF"

        else:
            return "Invalid Command"

# Main Program
agent = LightAgent()

while True:
    command = input("Enter command (ON/OFF/EXIT): ")

    if command == "EXIT":
        break

    print(agent.act(command))
```

```python
# Goal-Based Agent - GPS Navigation

current = input("Enter current location: ")
goal = input("Enter destination: ")

print("\nNavigation Started...")

while current != goal:
    print("Moving from", current, "to", goal)
    current = goal

print("Destination Reached!")
```

```python
# Goal-Based Agent using OOP

class GoalBasedAgent:
    def __init__(self, current_location, goal_location):
        self.current_location = current_location
        self.goal_location = goal_location

    def act(self):
        if self.current_location == self.goal_location:
            print("Goal already reached!")
        else:
            print(f"Moving from {self.current_location} to {self.goal_location}...")
            self.current_location = self.goal_location
            print("Package Delivered!")
            print("Goal Reached!")

# Main Program
current = input("Enter current location: ")
goal = input("Enter goal location: ")

agent = GoalBasedAgent(current, goal)
agent.act()
```

```python
# Utility-Based Agent using OOP

class UtilityAgent:
    def __init__(self, current_temperature):
        self.current_temperature = current_temperature

    def calculate_utility(self, temperature):
        # Higher utility means more comfort
        return 100 - abs(22 - temperature)

    def act(self):
        possible_temperatures = [18, 20, 22, 24, 26]

        best_temperature = possible_temperatures[0]
        max_utility = self.calculate_utility(best_temperature)

        for temp in possible_temperatures:
            utility = self.calculate_utility(temp)

            if utility > max_utility:
                max_utility = utility
                best_temperature = temp

        return best_temperature, max_utility


# Main Program
current_temp = int(input("Enter current temperature: "))

agent = UtilityAgent(current_temp)

temperature, utility = agent.act()

print("Best Temperature Setting:", temperature)
print("Utility Score:", utility)
```

```python
def simple_reflex_solar_agent(solar_output_kw, battery_percent):
    """Decides action purely on current inputs using hardcoded rules."""
    if solar_output_kw > 3.0 and battery_percent < 100:
        return "CHARGE_BATTERY"
    elif solar_output_kw > 0.0:
        return "POWER_HOME"
    else:
        return "DRAW_FROM_GRID"

# Example run
print(simple_reflex_solar_agent(solar_output_kw=4.5, battery_percent=60))
# Output: CHARGE_BATTERY
```
