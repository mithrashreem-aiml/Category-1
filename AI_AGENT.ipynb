```python
#SIMPLE REFLEX AGENT
class SimpleReflexAgent:
    def __init__(self):
        pass

    def act(self, location, status):
        """
        location: Current location ('A' or 'B')
        status: 'Dirty' or 'Clean'
        """
        if status == "Dirty":
            return "Suck"
        elif location == "A":
            return "Move Right"
        elif location == "B":
            return "Move Left"
agent = SimpleReflexAgent()

test_cases = [
    ("A", "Dirty"),
    ("A", "Clean"),
    ("B", "Dirty"),
    ("B", "Clean")
]

for location, status in test_cases:
    action = agent.act(location, status)
    print(f"Location: {location}, Status: {status} -> Action: {action}")
```

```python
# Model-Based Agent - Light Control

class LightAgent:
    def __init__(self):
        self.light_state = "OFF"   # Internal memory

    def act(self, command):
        if command == "ON":
            if self.light_state == "OFF":
                self.light_state = "ON"
                return "Light Turned ON"
            else:
                return "Light is already ON"

        elif command == "OFF":
            if self.light_state == "ON":
                self.light_state = "OFF"
                return "Light Turned OFF"
            else:
                return "Light is already OFF"

        else:
            return "Invalid Command"

# Main Program
agent = LightAgent()

while True:
    command = input("Enter command (ON/OFF/EXIT): ")

    if command == "EXIT":
        break

    print(agent.act(command))
```

```python
#goal based agent-Solar agent
solar = int(input("Solar generation (kW): "))
load = int(input("Current load demand (kW): "))
battery = int(input("Battery level (%): "))
weather = input("Weather (Sunny/Cloudy): ").lower()
price = input("Grid price (High/Low): ").lower()

print("\n------ Goal-Based Agent Decision ------")

if solar >= load:
    print("Power Source : Solar Panels")

    excess = solar - load

    if battery < 95:
        print("Action : Charge Battery")
    else:
        print("Action : Export", excess, "kW to Grid")

else:

    deficit = load - solar

    if battery > 30:
        print("Power Source : Solar + Battery")
        print("Battery supplies", deficit, "kW")

    else:

        if price == "low":
            print("Power Source : Grid Electricity")
            print("Battery reserved for emergencies")

        else:
            print("Grid price is high.")
            print("Delay washing machine, water heater, EV charging")
            print("Energy Saving Mode Activated")

if weather == "cloudy" and battery < 70:
    print("Recommendation : Charge battery during off-peak hours")

print("\nGoal Achieved:")
print("- Maximize Solar Utilization")
print("- Minimize Electricity Cost")
print("- Preserve Battery Health")
```

```python
#Utility based agents-Mindcare agent
actions = {
    "Meditation": 25,
    "Exercise": 20,
    "Sleep": 35,
    "Study": 10,
    "Social Interaction": 18,
    "Relaxation": 22
}

mood = int(input("Mood (0-100): "))
stress = int(input("Stress (0-100): "))
anxiety = int(input("Anxiety (0-100): "))
sleep = int(input("Sleep Quality (0-100): "))
energy = int(input("Energy (0-100): "))
workload = int(input("Workload (0-100): "))

best_action = ""
best_utility = -999

print("\nUtility Values")

for action, benefit in actions.items():

    utility = (
        0.30*mood
        +0.25*energy
        +0.20*sleep
        -0.15*stress
        -0.10*anxiety
        -0.05*workload
        +benefit
    )

    print(f"{action:20} = {utility:.2f}")

    if utility > best_utility:
        best_utility = utility
        best_action = action

print("\nBest Action :", best_action)
print("Maximum Utility :", round(best_utility,2))
```
